```python
###### ============================================
# Travel Expense Tracker
# Author: Reem Elshorbagy
# Description: A travel expense tracker
# tracker to log, categorize, and calculate
# ============================================
import pandas as pd

# List to define categories
categories = ['Food', 'Transportation', 'Accommodation', 'Activities']

# List to define cities
cities = []

# Dictionary to gather expenses
expenses = {'Food': {}, 'Transportation': {}, 'Accommodation': {}, 'Activities': {}}

# Function to add cities
def add_cities():
    while True:
        city = input("Please enter a city (or type 0 to exit): ").strip().title()
        
        if city == "0":
            break
        
        if city not in cities:
            cities.append(city)
            
            for category in expenses:
                expenses[category][city] = []
        
            print(f"{city} is added to the list.")
        
        else:
            print("This city is already added.")

# Function to add expenses
def add_expenses():
    while True:
        expense = input("Please enter the expense value (or type 0 to exit): ")
        
        if expense == "0":
            break
            
        try: 
            amount = float(expense)
        except ValueError:
            print("Please enter a valid number.")
            continue
        
        if amount <= 0:
            print("Please enter a positive value.")
            continue
        
        while True:
            category = input("Please enter the category (or type 0 to return to menu): ").strip().title()
            
            if category == "0":
                return
            
            if category in expenses:
                break

            print("Please choose from the pre-defined categories.")

        while True:
            city = input("Please enter the city: (or type 0 to return to menu)").strip().title()
            
            if city == "0":
                return

            if city in cities:
                break

            print("Please choose from the pre-defined cities.")

        expenses[category][city].append(amount)
        print(f"{amount} is added to {category}'s expenses in {city}.")

# Calculates the sum of expenses
def total_expenses():
    total = 0
    for category in expenses:
        for city in expenses[category]:
            total += sum(expenses[category][city])
    if total == 0:
        print("There are no expenses recorded")
    else:
        print(f"Total of expenses: {total}")

# Calculates the total for one specific category
def total_expenses_category():
    total = 0
    category = input("Please type the category name: ").strip().title()

    if category not in expenses:
        print("Invalid category.")
        return
        
    for city in expenses[category]:
        total += sum(expenses[category][city])
    
    print(f"Total expenses for {category}: {total}")

# Calculates the total for one specific city
def total_expenses_city():
    total = 0
    city = input("Please type the city name: ").strip().title()

    if city not in cities:
        print("Invalid city.")
        return

    for category in expenses:
        total += sum(expenses[category][city])
    print(f"The total of expenses in {city}: {total}.")

# Calculates the highest expense value
def highest(expenses):
    highest = 0
    for category in expenses:
        for city in expenses[category]:
            for element in expenses[category][city]:
                if element > highest:
                    highest = element
    print(f"The highest amount spent is {highest}.")

# Calculates the lowest expense value
def lowest(expenses):
    lowest = None
    for category in expenses:
        for city in expenses[category]:
            for element in expenses[category][city]:
                if lowest == None or element < lowest:
                    lowest = element
    print(f"The lowest amount spent is {lowest}.")

# Removes all entries
def clear_entries():
    global expenses
    
    expenses = {'Food': {}, 'Transportation': {}, 'Accommodation': {}, 'Activities': {}}

    cities.clear()
    print("All entries are cleared.")
    
    
# Menu options
def options():
    while True:
        number = input("Please pick from the options below:\n(1) To input a new city\n(2) To enter a new expense\n(3) To get the total of all expenses\n(4) To get the total of all expenses in a specific category\n(5) To get the total of all expenses in a specific city\n(6) To get the highest expense\n(7) To get the lowest expense\n(8) To display the logged expenses\n(9) To remove all entries\n(0) To break\nYour choice: ")
        
        if number == "0":
            break
        elif number == "1":
            add_cities()
        elif number == "2":
            add_expenses()
        elif number == "3":
            total_expenses()
        elif number == "4":
            total_expenses_category()
        elif number == "5":
            total_expenses_city()
        elif number == "6":
            highest(expenses)
        elif number == "7":
            lowest(expenses)
        elif number == "8":
            data = pd.DataFrame(expenses)
            for column in data.columns:
                data[column] = data[column].apply(
                    lambda values: "" if values == [] else values)
            display(data)
        elif number == "9":
            clear_entries()
        else:
            print("Please pick an option from 0 to 9.")

options()
        
```
